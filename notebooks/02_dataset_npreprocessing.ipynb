```python
gitignore_path = ".gitignore"
gitignore_text = """# Byte-compiled / optimized / DLL files
__pycache__/
*.py[cod]
*$py.class

# Virtual environment
.venv/
venv/
.env

# Jupyter
.ipynb_checkpoints/

# OS
.DS_Store
Thumbs.db

# Data
data/raw/*
!data/raw/.keep
data/external/*
!data/external/.keep

# Models & results
models/*
!models/.keep
results/*
!results/.keep

# Cache
*.cache
.cache/
"""

# Ensure keep files for empty dirs
for keep in ["data/raw/.keep", "data/external/.keep", "models/.keep", "results/.keep"]:
    os.makedirs(os.path.dirname(keep), exist_ok=True)
    if not os.path.exists(keep):
        with open(keep, "w", encoding="utf-8") as f:
            f.write("")

# Write .gitignore
with open(gitignore_path, "w", encoding="utf-8") as f:
    f.write(gitignore_text)

print("Wrote .gitignore and created .keep files.")
```
