# Synthetic Data Generation 🤖
**Objective:** Generate volunteer descriptions locally.
**Environment:** Local Repository Execution.

```python
import pandas as pd
import random
import os

# --- CONFIGURATION ---
BASE_DIR = r"E:\Github\repo-phaze7r\geospatial-tagging-thesis"
OUTPUT_FILE = os.path.join(BASE_DIR, 'datareported', 'synthetic_volunteered_descriptions.csv')
        
CITIES = ['Attock', 'Chitral', 'Islamabad', 'Karachi', 'Kohat', 'Lahore', 'Muzaffarabad', 'Peshawar', 'Quetta', 'Rawalpindi', 'Sialkot', 'Skardu']
print(f"[*] Base Directory: {BASE_DIR}")
print(f"[*] Output File: {OUTPUT_FILE}")
```

```python
TEMPLATES = [
    "I visited this {adj} {noun} and was amazed by the {adj2} {noun2}.",
    "A hidden gem! This {adj} {noun} features {adj2} {noun2}.",
    "Locals say this {adj} {noun} is the best spot for {adj2} {noun2}."
]
ADJECTIVES = ["beautiful", "historic", "modern", "famous", "local", "clean", "busy"]
OSM_TAGS = ["mosque", "shrine", "bank", "park", "shop", "hotel", "college"]
NOUNS_GENERIC = ["place", "spot", "view", "experience", "service"]

data = []
for i in range(500):
    city = random.choice(CITIES)
    template = random.choice(TEMPLATES)
    
    noun = random.choice(OSM_TAGS).replace('_', ' ')
    adj = random.choice(ADJECTIVES)
    noun2 = random.choice(OSM_TAGS + NOUNS_GENERIC).replace('_', ' ')
    adj2 = random.choice(ADJECTIVES)
    
    data.append({
        'id': i + 100000,
        'city': city,
        'role': 'Volunteer Guide',
        'synthetic_description': template.format(adj=adj, noun=noun, adj2=adj2, noun2=noun2)
    })

df = pd.DataFrame(data)
df.to_csv(OUTPUT_FILE, index=False)
print(f"[+] Saved: {OUTPUT_FILE}")
```
